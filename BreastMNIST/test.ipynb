```python
import os
import pickle
import numpy as np

model_dir = "saved_models"

specialists = [
    "BreastMNIST_Adaptive_Gaussian_Specialist.pkl",
    "BreastMNIST_Adaptive_Mean_Specialist.pkl",
    "BreastMNIST_Canny_Specialist.pkl",
    "BreastMNIST_Color_Thermometers_Specialist.pkl",
    "BreastMNIST_Otsu_Specialist.pkl",
]

for filename in specialists:
    model_path = os.path.join(model_dir, filename)
    with open(model_path, "rb") as f:
        tm = pickle.load(f)

    weights = tm.weight_banks[1].get_weights()
    n_pos  = np.sum(weights > 0)
    n_neg  = np.sum(weights < 0)
    n_zero = np.sum(weights == 0)
    half   = len(weights) // 2

    print(f"\n=== {filename} ===")
    print(f"Positive weights: {n_pos}")
    print(f"Negative weights: {n_neg}")
    print(f"Zero weights:     {n_zero}")
    print(f"Total clauses:    {len(weights)}")
    print(f"Expected half:    {half}")
    print(f"First half all positive: {np.all(weights[:half] > 0)}")
    print(f"Second half all negative: {np.all(weights[half:] < 0)}")
```
