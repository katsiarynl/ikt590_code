```python
import numpy as np
import os

# Load from both directories
data_dir_1 = "processed_data"
data_dir_2 = "processed_data_2"

files = {
    "X_train": "X_train_balanced.npy",
    "Y_train": "Y_train_balanced.npy",
    "X_test":  "X_test.npy",
    "Y_test":  "Y_test.npy",
    "X_val":   "X_val.npy",
    "Y_val":   "Y_val.npy",
}

all_match = True
for name, filename in files.items():
    a = np.load(os.path.join(data_dir_1, filename))
    b = np.load(os.path.join(data_dir_2, filename))
    
    shape_match = a.shape == b.shape
    values_match = np.array_equal(a, b)
    
    status = "✓ MATCH" if (shape_match and values_match) else "✗ MISMATCH"
    print(f"{name}: {status} | shape: {a.shape} vs {b.shape} | values equal: {values_match}")
    
    if not (shape_match and values_match):
        all_match = False

print("\n" + ("All datasets are identical." if all_match else "Datasets differ."))
```
