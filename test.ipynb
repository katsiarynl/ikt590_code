```python
import matplotlib.pyplot as plt
import numpy as np
from medmnist import BloodMNIST

dataset_28 = BloodMNIST(split="val", download=True, size=28)
dataset_128 = BloodMNIST(split="val", download=True, size=128)

img_28, _ = dataset_28[0]
img_128, _ = dataset_128[0]

plt.figure(figsize=(8,4))

plt.subplot(1,2,1)
plt.imshow(np.array(img_28), cmap='gray')
plt.title("28 × 28 (original)")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(np.array(img_128), cmap='gray')
plt.title("128 × 128 (resized)")
plt.axis('off')

plt.show()
```

```python
import numpy as np
from scipy.stats import mannwhitneyu

# positive clauses clearly longer than negative
data_pos = {'test': np.array([100, 110, 120])}
data_neg = {'test': np.array([1, 2, 3])}

stat, p = mannwhitneyu(data_pos['test'], data_neg['test'], alternative='two-sided')
n1, n2 = len(data_pos['test']), len(data_neg['test'])
r = 1 - (2 * stat) / (n1 * n2)
print(f"r = {r:.4f}")  # should be negative if positive > negative
```
